```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from io import BytesIO
from PIL import Image

uploaded = files.upload()

img = np.array(Image.open(BytesIO(next(iter(uploaded.values())))).convert('L'))

negative = 255 - img

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(negative, cmap='gray')
plt.title('Negative Image')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from io import BytesIO
from PIL import Image


uploaded = files.upload()

img = np.array(Image.open(BytesIO(next(iter(uploaded.values())))).convert('L'))

gamma = 0.5
normalized = img / 255.0
power_law = np.uint8(255 * np.power(normalized, gamma))

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(power_law, cmap='gray')
plt.title('Power Law Transformed Image')
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
from io import BytesIO
from PIL import Image

uploaded = files.upload()

img = np.array(Image.open(BytesIO(next(iter(uploaded.values())))).convert('L'))

c = 255 / np.log(1 + np.max(img))
log_transform = c * np.log(1 + img)
log_transform = np.uint8(log_transform)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(log_transform, cmap='gray')
plt.title('Log Transformed Image')
plt.axis('off')

plt.show()
```
