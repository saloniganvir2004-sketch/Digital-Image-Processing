Assignment 1

Write a program to load a colour image from corel -1k dataset and convert it into gray scale. Perform the following operations.
i) Obtain negative transform
ii) Obtain log transform
iii) Obtain power law transform
Repeat these operations on 5 different images
Steps to follow:
1. Load a colour image
2. Convert it into grayscale
3. Access each element/number/gray scale intensity value
4. Apply negative transform s=L-1-r (where s is intensity value of transformed image, L is 256 and intensity value of original image)
5. Store each value of s in a matrix
6. Display he matrix as image

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

image_names = list(uploaded.keys())

print("Number of images loaded:", len(image_names))
print("Image names:", image_names)
```

```python
# Step 2: Convert all images to grayscale

gray_images = []

for name in image_names:
    image = cv2.imread(name)

    # Convert colour image to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    gray_images.append(gray)

plt.figure(figsize=(15, 8))

for i, gray in enumerate(gray_images):

    plt.subplot(2, 3, i + 1)

    plt.imshow(gray, cmap='gray')

    plt.title("Grayscale Image " + str(i + 1))

    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Step 3: Access each element and obtain its grayscale intensity

for k in range(len(gray_images)):

    gray = gray_images[k]

    rows, columns = gray.shape

    print("\nImage", k + 1)
    print("Number of rows:", rows)
    print("Number of columns:", columns)

    for i in range(rows):
        for j in range(columns):

            r = gray[i][j]

            if i < 2 and j < 5:
                print("Pixel [", i, "][", j, "] =", r)
```

```python
# Step 4: Apply Negative Transformation
# Formula: s = L - 1 - r

negative_images = []

for k in range(len(gray_images)):

    gray = gray_images[k]

    rows, columns = gray.shape

    negative = np.zeros((rows, columns), dtype=np.uint8)

    for i in range(rows):
        for j in range(columns):

            r = gray[i][j]

            L = 256

            s = (L - 1) - r

            negative[i][j] = s

    negative_images.append(negative)

plt.figure(figsize=(15, 8))

for i, negative in enumerate(negative_images):

    plt.subplot(2, 3, i + 1)

    plt.imshow(negative, cmap='gray')

    plt.title("Negative Image " + str(i + 1))

    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Step 6: Display the negative matrix as an image

for k in range(len(negative_images)):

    plt.figure(figsize=(6, 5))

    plt.imshow(negative_images[k], cmap='gray')

    plt.title("Negative Image " + str(k + 1))

    plt.axis("off")

    plt.show()
```

```python
# Display original, grayscale and negative images

for k in range(len(image_names)):

    original = cv2.imread(image_names[k])
    original = cv2.cvtColor(original, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(original)
    plt.title("Original Colour Image")
    plt.axis("off")

    plt.subplot(1, 3, 2)
    plt.imshow(gray_images[k], cmap='gray')
    plt.title("Grayscale Image")
    plt.axis("off")

    plt.subplot(1, 3, 3)
    plt.imshow(negative_images[k], cmap='gray')
    plt.title("Negative Image")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```

```python
# Log Transformation
# Formula: s = c * log(1 + r)
# c = 255 / log(256)

log_images = []

c = 255 / np.log(256)

for k in range(len(gray_images)):
    gray = gray_images[k]
    rows, columns = gray.shape
    log_matrix = np.zeros((rows, columns), dtype=np.uint8)

    for i in range(rows):
        for j in range(columns):
            r = gray[i][j]
            s = c * np.log(1 + r)
            log_matrix[i][j] = np.uint8(s)

    log_images.append(log_matrix)

plt.figure(figsize=(15, 8))

for i, log_image in enumerate(log_images):
    plt.subplot(2, 3, i + 1)
    plt.imshow(log_image, cmap='gray')
    plt.title("Log Transform Image " + str(i + 1))
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Power-Law Transformation
# Formula: s = c * r^gamma

power_images = []
c = 1
gamma = 0.9

for k in range(len(gray_images)):

    gray = gray_images[k]

    rows, columns = gray.shape
    power_matrix = np.zeros((rows, columns), dtype=np.uint8)

    for i in range(rows):
        for j in range(columns):
            r = gray[i][j]
            r_normalized = r / 255.0
            s = c * (r_normalized ** gamma)
            s = s * 255
            power_matrix[i][j] = np.uint8(s)
    power_images.append(power_matrix)

plt.figure(figsize=(15, 8))

for i, power_image in enumerate(power_images):

    plt.subplot(2, 3, i + 1)

    plt.imshow(power_image, cmap='gray')

    plt.title("Power Law Image " + str(i + 1))

    plt.axis("off")

plt.tight_layout()
plt.show()
```
